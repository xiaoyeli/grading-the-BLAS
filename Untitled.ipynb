```python
import h5py
import xarray as xr
import numpy as np
```

```python
h5_file = h5py.File("grading_dgemm.h5", "r")
h5_dims = list(h5_file.keys())

data = []
coords = {'implementation': [], 'matrix_type': [], 'dimension': [], 'seed': []}

for mat_type in h5_file.keys():
    for dim in h5_file[mat_type].keys():
        for seed in h5_file[mat_type][dim].keys():
            for impl in h5_file[mat_type][dim][seed].keys():
                if impl == 'A' or impl == 'B':
                    continue
                data.append(h5_file[mat_type][dim][seed][impl]['C'][None, ...])
                coords['implementation'].append("_".join(impl.split("_")[1:]))
                coords['matrix_type'].append("_".join(mat_type.split("_")[1:]))
                coords['seed'].append(int(seed.split("_")[1]))
                coords['dimension'].append("_".join(dim.split("_")[1:]))
```

```python
data = np.concatenate(data, axis=0)  # shape: (n_samples, n, n)
ds = xr.DataArray(
    data,
    dims=("sample", "i", "j"),
    coords={
        "implementation": ("sample", coords['implementation']),
        "matrix_type": ("sample", coords['matrix_type']),
        "seed": ("sample", coords['seed']),
        "dimension": ("sample", coords['dimension']),
    },
)
```
